```python
import pytesseract
from PIL import Image
import os

# Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')
```

Load Image & OCR Extraction

```python
# Define the path to your signature image
path = '/content/drive/MyDrive/Rana Jadu.png'

extracted_text = ""  # IMPORTANT: initialize to avoid runtime error

try:
    img = Image.open(path)
    extracted = pytesseract.image_to_string(img)
    extracted_text = extracted.strip()

    print(f"OCR Result: '{extracted_text}'")
except FileNotFoundError:
    print("Error: Image file not found at the specified path.")
```

️Knowledge Factor (OCR vs User Input)

```python
entered_data = input("Enter your signature (Type the name): ")

if extracted_text.lower() == entered_data.lower():
    print("Result: ACCESS GRANTED (Match found)")
else:
    print("Result: ACCESS DENIED")
```

Simulated 2FA Data

```python
PREDEFINED_PASSWORD = "StrongSecurePassword123"
PREDEFINED_TOTP_CODE = "123456"
```

 2FA Functions

```python
def check_knowledge_factor():
    print("\n--- FACTOR 1: KNOWLEDGE (Password) ---")
    entered_password = input("Enter your secure Vault Password: ")

    if entered_password == PREDEFINED_PASSWORD:
        print(" Password check successful.")
        return True
    else:
        print(" Password incorrect.")
        return False
```

```python
def check_possession_factor():
    print("\n--- FACTOR 2: POSSESSION (Authenticator Code) ---")
    entered_code = input("Enter the 6-digit code from your Authenticator App: ")

    if entered_code == PREDEFINED_TOTP_CODE:
        print(" TOTP Code check successful.")
        return True
    else:
        print(" TOTP Code incorrect.")
        return False
```

2FA Execution Logic


```python
print("--- Starting Two-Factor Authentication (2FA) for Vault ---")

password_ok = check_knowledge_factor()

if password_ok:
    code_ok = check_possession_factor()

    if code_ok:
        print("\n*** ACCESS GRANTED ***")
        print("Both factors verified. Welcome to the Vault Application.")
    else:
        print("\n*** ACCESS DENIED ***")
        print("Possession factor failed. Access denied.")
else:
    print("\n*** ACCESS DENIED ***")
    print("Knowledge factor failed. Access denied.")
```
